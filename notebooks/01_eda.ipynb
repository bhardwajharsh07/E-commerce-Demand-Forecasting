### Converting Date Columns

The date columns are currently stored as 'str'. They need to be converted into datetime format before analyzing demand over time.

```python
ecom_df["order date (DateOrders)"].min(), ecom_df["order date (DateOrders)"].max()
```

The order data covers multiple years, which gives us enough historical information to study demand trends over time.

The date columns have now been converted from 'str' to datetime, making them suitable for time-based aggregation and forecasting.


### Daily Demand Trend

```python
daily_demand = (
    ecom_df.groupby(
        ecom_df["order date (DateOrders)"].dt.date
    )["Order Item Quantity"]
    .sum()
)

plt.figure(figsize=(14, 6))
daily_demand.plot(color="yellow")
plt.title("Daily E-commerce Demand")
plt.xlabel("Date")
plt.ylabel("Total Units Demanded")
plt.show()
```

The daily demand shows noticeable fluctuations over time rather than remaining constant.

There are several peaks and drops in demand, indicating that demand is not completely stable and that time-based patterns may be useful for forecasting.


### Monthly Demand Trend

```python
monthly_demand = (
    ecom_df.set_index("order date (DateOrders)")
    .resample("ME")["Order Item Quantity"]
    .sum()
)

plt.figure(figsize=(14, 6))
monthly_demand.plot(color="yellow")
plt.title("Monthly E-commerce Demand")
plt.xlabel("Month")
plt.ylabel("Total Units Demanded")
plt.show()
```

Monthly aggregation makes the overall demand pattern easier to observe.

The demand varies across months, suggesting that monthly and other calendar-based features may be useful in the later forecasting model.

### Preparing Data for Baseline Model

```python
daily_df = (
    ecom_df.groupby(
        pd.Grouper(
            key="order date (DateOrders)",
            freq="D"
        )
    )["Order Item Quantity"]
    .sum()
    .reset_index()
)

daily_df.columns = ["date", "demand"]

daily_df["lag_1"] = daily_df["demand"].shift(1)

daily_df = daily_df.dropna().copy()

daily_df.head()
```

For the baseline model, the daily order quantity is used as the demand target.

The previous day's demand is used as the first lag feature because it is available before the current day's demand and therefore does not introduce future information into the model.


### Baseline Model

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error

split_index = int(len(daily_df) * 0.8)

train = daily_df.iloc[:split_index].copy()
test = daily_df.iloc[split_index:].copy()

X_train = train[["lag_1"]]
y_train = train["demand"]

X_test = test[["lag_1"]]
y_test = test["demand"]

model = LinearRegression()
model.fit(X_train, y_train)

predictions = model.predict(X_test)

mae = mean_absolute_error(y_test, predictions)
rmse = np.sqrt(mean_squared_error(y_test, predictions))

print("Baseline MAE:", mae)
print("Baseline RMSE:", rmse)
```

The baseline model uses Linear Regression with only the previous day's demand as the input feature.

The chronological 80%-20% split keeps earlier observations in the training set and later observations in the test set, which is more appropriate for a forecasting problem than a random split.


### Baseline Forecast vs Actual Demand

```python
plt.figure(figsize=(14, 6))

plt.plot(
    test["date"],
    y_test,
    label="Actual Demand",
    color="white"
)

plt.plot(
    test["date"],
    predictions,
    label="Baseline Prediction",
    color="yellow"
)

plt.title("Baseline Forecast vs Actual Demand")
plt.xlabel("Date")
plt.ylabel("Demand")
plt.legend()
plt.show()
```

The baseline predictions follow the general movement of the actual demand but do not capture every demand spike and drop.

This gives us a benchmark that future forecasting models should improve upon.

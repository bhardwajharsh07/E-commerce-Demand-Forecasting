```python
product_demand = (
    ecom_df.groupby("Product Name")["Order Item Quantity"].sum().sort_values(ascending = False)
)

product_demand.head(10)
```

```python
plt.figure(figsize=(10, 6))
product_demand.head(10).sort_values().plot(kind="barh")
plt.title("Top 10 Products by Total Demand")
plt.xlabel("Total Units Sold")
plt.ylabel("Product")
plt.show()
```
